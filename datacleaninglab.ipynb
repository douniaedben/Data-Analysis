```python

missing_columns = missing_values[missing_values > 0]
print("Columns with Missing Values:")
print(missing_columns)
```

```python

dirty_data['Customer_Name'].fillna('Unknown', inplace=True)

# Drop rows
dirty_data = dirty_data.dropna(subset=['Customer_Email'])

dirty_data['Transaction_Amount'] = pd.to_numeric(dirty_data['Transaction_Amount'], errors='coerce')  # Convert to numeric
median_amount = dirty_data['Transaction_Amount'].median()
dirty_data['Transaction_Amount'].fillna(median_amount, inplace=True)

# Drop rows
dirty_data = dirty_data.dropna(subset=['Product_Category', 'Transaction_Date'])
dirty_data.info()
dirty_data.sample(10)
```

```python
# date to standard date format
dirty_data['Transaction_Date'] = pd.to_datetime(dirty_data['Transaction_Date'], errors='coerce')
date_columns = dirty_data.select_dtypes(include=['datetime']).columns
print("Date Columns:")
print(date_columns)

#  Product_Category to lowercase
dirty_data['Product_Category'] = dirty_data['Product_Category'].str.lower()

# Check for non-numeric values in 'Transaction_Amount' and handle them
dirty_data['Transaction_Amount'] = pd.to_numeric(dirty_data['Transaction_Amount'], errors='coerce')

dirty_data.sample(5)
```

```python
# Identify all columns with text data object type columns
text_columns = dirty_data.select_dtypes(include=['object']).columns

#  Capitalizing first letters
for col in text_columns:
    dirty_data[col] = dirty_data[col].str.title()

# Check the updated dataset
print(dirty_data.head())
```

```python



# Detect and cap outliers in 'Transaction_Amount' using IQR ....> cour
Q1 = dirty_data['Transaction_Amount'].quantile(0.25)
Q3 = dirty_data['Transaction_Amount'].quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Cap values outside the range
dirty_data['Transaction_Amount'] = dirty_data['Transaction_Amount'].clip(lower=lower_limit, upper=upper_limit)

dirty_data.sample(5)
```

```python
# Remove duplicate rows
dirty_data = dirty_data.drop_duplicates()

# Check duplicates removal
print(f"Dataset Shape after removing duplicates: {dirty_data.shape}")
```

```python
dirty_data.duplicated().sum()



```

```python
# Standardize 'Product_Category' values and fix common typos
category_corrections = {
    'elecronics': 'electronics',
    'fashiion': 'fashion',
    'hoome': 'home',
     'health': 'health & beauty',
    'beauty': 'health & beauty'
}

dirty_data['Product_Category'] = dirty_data['Product_Category'].replace(category_corrections)
dirty_data['Product_Category'] = dirty_data['Product_Category'].str.lower()

print(dirty_data['Product_Category'].unique())
```

```python
# Display final dataset shape and save to a new file
print(f"Final Dataset Shape: {dirty_data.shape}")
dirty_data.to_csv('/content/drive/MyDrive/cleaned_ecommerce_transactions.csv', index=False)

dirty_data.sample(5)
```
